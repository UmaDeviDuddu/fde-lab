```python
repos = [
    {"name":"web-portal", "size": 700, "uses_lfs":False},
    {"name":"api-server", "size": 500, "uses_lfs":True},
    {"name":"data-pipeline","size": 1500, "uses_lfs":True},
    {"name":"mobile-app", "size": 200, "uses_lfs":True}

]

repos[2]["name"], len(repos)
```

```python
for r in repos:
    print(f"{r['name']}: {r['size']} MB, uses LFS: {r['uses_lfs']}")
    
```

```python
counts = {}
for r in repos:
    lfs = "LFS" if r["uses_lfs"] else "no LFS"
    counts[lfs] = counts.get(lfs, 0) + 1

counts
```
